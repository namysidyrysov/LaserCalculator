```python
import tkinter as tk

root = tk.Tk()

var = tk.StringVar()
var.set("Привет")

entry = tk.Entry(root, textvariable=var)
entry.pack()

label = tk.Label(root, textvariable=var)
label.pack()

def show():
    print(var.get())

btn = tk.Button(root, text="Показать", command=show)
btn.pack()

root.mainloop()
```
